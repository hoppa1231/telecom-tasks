### Упражнение 4.1: Типы шума

Цель: познакомиться с различными типами шума (белый, розовый, коричневый) и их спектральными характеристиками.

```python
# Импорт необходимых библиотек
import sys
sys.path.insert(0, '../ThinkDSP/code')

from thinkdsp import Signal, UncorrelatedGaussianNoise, BrownianNoise, PinkNoise
from thinkdsp import decorate
import matplotlib.pyplot as plt
import numpy as np
```

### 1. Белый шум (White Noise)

Белый шум имеет равномерное распределение энергии по всем частотам.

```python
# Создание белого шума
signal = UncorrelatedGaussianNoise()
wave = signal.make_wave(duration=1.0, framerate=11025)

# Визуализация временной области
segment = wave.segment(duration=0.1)
segment.plot()
decorate(xlabel='Время (с)', ylabel='Амплитуда')
plt.show()
```

```python
# Спектр белого шума
spectrum = wave.make_spectrum()
spectrum.plot_power()
decorate(xlabel='Частота (Гц)', ylabel='Мощность')
plt.show()
```

**Вопрос 1:** Как выглядит спектр белого шума? Почему он называется "белым"?

### 2. Розовый шум (Pink Noise)

Розовый шум имеет спектральную плотность мощности, обратно пропорциональную частоте (1/f).

```python
# Создание розового шума
signal = PinkNoise(beta=1)
wave = signal.make_wave(duration=1.0, framerate=11025)

# Визуализация
segment = wave.segment(duration=0.1)
segment.plot()
decorate(xlabel='Время (с)', ylabel='Амплитуда')
plt.show()
```

```python
# Спектр розового шума
spectrum = wave.make_spectrum()
spectrum.plot_power()
decorate(xlabel='Частота (Гц)', ylabel='Мощность')
plt.show()
```

**Вопрос 2:** Чем отличается спектр розового шума от белого? Какой наклон имеет спектр на логарифмической шкале?

### 3. Коричневый (Броуновский) шум

Коричневый шум имеет спектральную плотность мощности 1/f².

```python
# Создание коричневого шума
signal = BrownianNoise()
wave = signal.make_wave(duration=1.0, framerate=11025)

# Визуализация
segment = wave.segment(duration=0.1)
segment.plot()
decorate(xlabel='Время (с)', ylabel='Амплитуда')
plt.show()
```

```python
# Спектр коричневого шума
spectrum = wave.make_spectrum()
spectrum.plot_power()
decorate(xlabel='Частота (Гц)', ylabel='Мощность')
plt.show()
```

**Вопрос 3:** Как меняется спектр при переходе от белого к розовому и коричневому шуму?

### 4. Сравнение типов шума

Создадим все три типа шума и сравним их спектры на одном графике.

```python
# Создание всех типов шума
duration = 1.0
framerate = 11025

white = UncorrelatedGaussianNoise().make_wave(duration, framerate)
pink = PinkNoise(beta=1).make_wave(duration, framerate)
brown = BrownianNoise().make_wave(duration, framerate)

# Построение спектров
white_spectrum = white.make_spectrum()
pink_spectrum = pink.make_spectrum()
brown_spectrum = brown.make_spectrum()

# Визуализация на одном графике
plt.figure(figsize=(10, 6))
white_spectrum.plot_power(label='Белый шум', alpha=0.7)
pink_spectrum.plot_power(label='Розовый шум', alpha=0.7)
brown_spectrum.plot_power(label='Коричневый шум', alpha=0.7)
plt.legend()
decorate(xlabel='Частота (Гц)', ylabel='Мощность')
plt.show()
```
