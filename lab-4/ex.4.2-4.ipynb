### Упражнение 4.2: Низкочастотные и высокочастотные фильтры

В этом упражнении мы изучим применение фильтров для обработки сигналов.

```python
# Импорт библиотек
import sys
sys.path.insert(0, '../ThinkDSP/code')

from thinkdsp import read_wave, SinSignal, SquareSignal, TriangleSignal
from thinkdsp import decorate
import matplotlib.pyplot as plt
import numpy as np
```

### 1. Низкочастотный фильтр (Low-pass filter)

Низкочастотный фильтр пропускает низкие частоты и ослабляет высокие.

```python
# Создание сигнала с несколькими гармониками
signal = SquareSignal(freq=440)
wave = signal.make_wave(duration=0.5, framerate=44100)

# Визуализация исходного сигнала
segment = wave.segment(duration=0.01)
segment.plot()
decorate(xlabel='Время (с)', ylabel='Амплитуда', title='Исходный сигнал')
plt.show()
```

```python
# Спектр исходного сигнала
spectrum = wave.make_spectrum()
spectrum.plot()
decorate(xlabel='Частота (Гц)', ylabel='Амплитуда', title='Спектр исходного сигнала')
plt.show()
```

```python
# Применение низкочастотного фильтра
cutoff = 2000  # Частота среза 2 кГц
spectrum.low_pass(cutoff)

# Спектр после фильтрации
spectrum.plot()
decorate(xlabel='Частота (Гц)', ylabel='Амплитуда', 
         title=f'Спектр после НЧ фильтра (cutoff={cutoff} Гц)')
plt.show()
```

```python
# Восстановление сигнала из отфильтрованного спектра
filtered_wave = spectrum.make_wave()
filtered_segment = filtered_wave.segment(duration=0.01)
filtered_segment.plot()
decorate(xlabel='Время (с)', ylabel='Амплитуда', title='Отфильтрованный сигнал')
plt.show()
```

### 2. Высокочастотный фильтр (High-pass filter)

Высокочастотный фильтр пропускает высокие частоты и ослабляет низкие.

```python
# Создание нового сигнала
signal = TriangleSignal(freq=440)
wave = signal.make_wave(duration=0.5, framerate=44100)
spectrum = wave.make_spectrum()

# Применение высокочастотного фильтра
cutoff = 1000  # Частота среза 1 кГц
spectrum.high_pass(cutoff)

# Визуализация
spectrum.plot()
decorate(xlabel='Частота (Гц)', ylabel='Амплитуда',
         title=f'Спектр после ВЧ фильтра (cutoff={cutoff} Гц)')
plt.show()
```

```python
# Восстановление и визуализация
filtered_wave = spectrum.make_wave()
filtered_segment = filtered_wave.segment(duration=0.01)
filtered_segment.plot()
decorate(xlabel='Время (с)', ylabel='Амплитуда', title='Сигнал после ВЧ фильтра')
plt.show()
```

### Упражнение 4.3: Полосовой фильтр (Band-pass filter)

Полосовой фильтр пропускает только определенный диапазон частот.

```python
# Создание сложного сигнала (сумма нескольких частот)
sig1 = SinSignal(freq=440)  # Ля первой октавы
sig2 = SinSignal(freq=880)  # Ля второй октавы
sig3 = SinSignal(freq=1760) # Ля третьей октавы

signal = sig1 + sig2 + sig3
wave = signal.make_wave(duration=0.5, framerate=44100)

# Спектр исходного сигнала
spectrum = wave.make_spectrum()
spectrum.plot(high=3000)
decorate(xlabel='Частота (Гц)', ylabel='Амплитуда', title='Исходный спектр')
plt.show()
```

```python
# Применение полосового фильтра
low_cutoff = 600
high_cutoff = 1200

spectrum.band_stop(low_cutoff, high_cutoff)

# Визуализация отфильтрованного спектра
spectrum.plot(high=3000)
decorate(xlabel='Частота (Гц)', ylabel='Амплитуда',
         title=f'Спектр после полосового фильтра ({low_cutoff}-{high_cutoff} Гц)')
plt.show()
```

### Упражнение 4.4: Фильтрация реального аудио

Применим фильтры к реальной записи звука.

```python
# Загрузка аудиофайла
wave = read_wave('../lab-1/92002__jcveliz__violin-origional.wav')

# Выделение сегмента
segment = wave.segment(start=1.2, duration=0.6)
segment.normalize()

# Визуализация
segment.plot()
decorate(xlabel='Время (с)', ylabel='Амплитуда', title='Исходная запись')
plt.show()
```

```python
# Спектр исходной записи
spectrum = segment.make_spectrum()
spectrum.plot(high=10000)
decorate(xlabel='Частота (Гц)', ylabel='Амплитуда', title='Спектр исходной записи')
plt.show()
```

```python
# Применение низкочастотного фильтра для удаления высокочастотного шума
spectrum_filtered = segment.make_spectrum()
spectrum_filtered.low_pass(3000)

# Визуализация отфильтрованного спектра
spectrum_filtered.plot(high=10000)
decorate(xlabel='Частота (Гц)', ylabel='Амплитуда', 
         title='Спектр после НЧ фильтра (3 кГц)')
plt.show()
```

```python
# Восстановление отфильтрованного сигнала
filtered_wave = spectrum_filtered.make_wave()
filtered_wave.normalize()
filtered_wave.apodize()

# Визуализация
filtered_wave.plot()
decorate(xlabel='Время (с)', ylabel='Амплитуда', title='Отфильтрованный сигнал')
plt.show()
```

```python
# Прослушивание исходного сигнала
print("Исходный сигнал:")
segment.apodize()
segment.make_audio()
```

```python
# Прослушивание отфильтрованного сигнала
print("Отфильтрованный сигнал:")
filtered_wave.make_audio()
```

### Дополнительное задание: Создание эквалайзера

Создайте простой эквалайзер, который позволяет усиливать или ослаблять различные частотные диапазоны.

```python
# Пример кода для эквалайзера
def simple_equalizer(wave, bands):
    """
    Простой эквалайзер
    
    wave: исходный сигнал
    bands: список кортежей (low_freq, high_freq, gain)
           где gain - коэффициент усиления (1.0 = без изменений)
    """
    spectrum = wave.make_spectrum()
    
    for low_freq, high_freq, gain in bands:
        # Найти индексы частот в спектре
        low_idx = np.searchsorted(spectrum.fs, low_freq)
        high_idx = np.searchsorted(spectrum.fs, high_freq)
        
        # Применить усиление
        spectrum.hs[low_idx:high_idx] *= gain
    
    return spectrum.make_wave()

# Пример использования
# bands = [(0, 500, 1.5),      # Усилить басы
#          (500, 2000, 1.0),   # Средние частоты без изменений
#          (2000, 10000, 0.5)] # Ослабить высокие частоты
# 
# equalized = simple_equalizer(wave, bands)
```
