## Лабораторная работа 5: Автокорреляция

### Упражнение 5.1: Основы автокорреляции

Автокорреляция - это мера сходства сигнала с самим собой при различных временных сдвигах.

```python
import sys
sys.path.insert(0, '../ThinkDSP/code')

from thinkdsp import SinSignal, SquareSignal, TriangleSignal
from thinkdsp import decorate
import matplotlib.pyplot as plt
import numpy as np
```

### 1. Автокорреляция синусоидального сигнала

```python
# Создание синусоидального сигнала
signal = SinSignal(freq=440)
wave = signal.make_wave(duration=0.5, framerate=10000)

# Визуализация сигнала
segment = wave.segment(duration=0.01)
segment.plot()
decorate(xlabel='Время (с)', ylabel='Амплитуда')
plt.show()
```

```python
# Вычисление автокорреляции
corr = np.correlate(wave.ys, wave.ys, mode='same')

# Нормализация
corr = corr / corr.max()

# Визуализация
N = len(corr)
lags = np.arange(-N//2, N//2)
plt.plot(lags, corr)
decorate(xlabel='Сдвиг (отсчеты)', ylabel='Автокорреляция')
plt.show()
```

### 2. Определение периода сигнала

```python
# Функция для поиска периода
def estimate_period(wave):
    """
    Оценка периода сигнала по автокорреляции
    """
    corr = np.correlate(wave.ys, wave.ys, mode='same')
    corr = corr / corr.max()
    
    # Ищем первый максимум после центра
    N = len(corr)
    center = N // 2
    
    # Пропускаем центральный пик
    start = center + 10
    
    # Находим следующий максимум
    peaks = []
    for i in range(start, N-1):
        if corr[i] > corr[i-1] and corr[i] > corr[i+1]:
            peaks.append(i)
    
    if peaks:
        period_samples = peaks[0] - center
        period_seconds = period_samples / wave.framerate
        frequency = 1 / period_seconds
        return period_samples, period_seconds, frequency
    return None, None, None

# Применение к нашему сигналу
period_samples, period_seconds, frequency = estimate_period(wave)
print(f"Период: {period_samples} отсчетов")
print(f"Период: {period_seconds:.6f} секунд")
print(f"Частота: {frequency:.2f} Гц")
print(f"Исходная частота: 440 Гц")
```

### 3. Автокорреляция различных сигналов

```python
# Создание различных сигналов
signals = [
    ('Синус', SinSignal(freq=440)),
    ('Прямоугольник', SquareSignal(freq=440)),
    ('Треугольник', TriangleSignal(freq=440))
]

fig, axes = plt.subplots(3, 2, figsize=(12, 10))

for i, (name, signal) in enumerate(signals):
    wave = signal.make_wave(duration=0.5, framerate=10000)
    
    # Сигнал
    segment = wave.segment(duration=0.01)
    axes[i, 0].plot(segment.ts, segment.ys)
    axes[i, 0].set_title(f'{name} - Сигнал')
    axes[i, 0].set_xlabel('Время (с)')
    axes[i, 0].set_ylabel('Амплитуда')
    
    # Автокорреляция
    corr = np.correlate(wave.ys, wave.ys, mode='same')
    corr = corr / corr.max()
    N = len(corr)
    lags = np.arange(-N//2, N//2)
    axes[i, 1].plot(lags[:1000], corr[N//2:N//2+1000])
    axes[i, 1].set_title(f'{name} - Автокорреляция')
    axes[i, 1].set_xlabel('Сдвиг (отсчеты)')
    axes[i, 1].set_ylabel('Корреляция')

plt.tight_layout()
plt.show()
```
