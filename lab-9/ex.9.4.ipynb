## Лабораторная работа 9: Дифференцирование и интегрирование

### Упражнение 9.1: Операции diff и differentiate

```python
import sys
sys.path.insert(0, '../ThinkDSP/code')

from thinkdsp import TriangleSignal
from thinkdsp import decorate
import matplotlib.pyplot as plt
import numpy as np
```

### Исходный треугольный сигнал

```python
# Создание треугольного сигнала
in_wave = TriangleSignal(freq=50).make_wave(duration=0.1, framerate=44100)

in_wave.plot()
plt.title('Исходный треугольный сигнал')
decorate(xlabel='Время (с)', ylabel='Амплитуда')
plt.show()
```

### Применение операции diff

Операция `diff` вычисляет разность между соседними отсчетами.

```python
# Применение diff
out_wave_diff = in_wave.diff()

out_wave_diff.plot()
plt.title('Результат операции diff')
decorate(xlabel='Время (с)', ylabel='Амплитуда')
plt.show()
```

**Комментарий:** Производная треугольного сигнала - это прямоугольный сигнал. Это объясняет, почему гармоники прямоугольного сигнала убывают как 1/f, в то время как у треугольного - как 1/f².

### Применение операции differentiate

Операция `differentiate` работает в частотной области.

```python
# Применение differentiate через спектр
spectrum = in_wave.make_spectrum()
spectrum_diff = spectrum.differentiate()
out_wave_spectral = spectrum_diff.make_wave()

out_wave_spectral.plot()
plt.title('Результат операции differentiate')
decorate(xlabel='Время (с)', ylabel='Амплитуда')
plt.show()
```

**Комментарий:** При спектральном дифференцировании возникает эффект "звона" (ringing) вокруг разрывов. Математически это связано с тем, что производная треугольного сигнала не определена в точках излома.

### Сравнение методов

```python
# Сравнение обоих методов
fig, axes = plt.subplots(3, 1, figsize=(12, 10))

# Исходный сигнал
axes[0].plot(in_wave.ts, in_wave.ys)
axes[0].set_title('Исходный треугольный сигнал')
axes[0].set_xlabel('Время (с)')
axes[0].set_ylabel('Амплитуда')
axes[0].grid(True, alpha=0.3)

# diff
axes[1].plot(out_wave_diff.ts, out_wave_diff.ys)
axes[1].set_title('Результат diff (временная область)')
axes[1].set_xlabel('Время (с)')
axes[1].set_ylabel('Амплитуда')
axes[1].grid(True, alpha=0.3)

# differentiate
axes[2].plot(out_wave_spectral.ts, out_wave_spectral.ys)
axes[2].set_title('Результат differentiate (частотная область)')
axes[2].set_xlabel('Время (с)')
axes[2].set_ylabel('Амплитуда')
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### Анализ спектров

```python
# Сравнение спектров
fig, axes = plt.subplots(2, 1, figsize=(12, 8))

# Спектр исходного сигнала
spectrum_orig = in_wave.make_spectrum()
spectrum_orig.plot(high=1000, ax=axes[0])
axes[0].set_title('Спектр исходного сигнала')
axes[0].set_xlabel('Частота (Гц)')
axes[0].set_ylabel('Амплитуда')
axes[0].grid(True, alpha=0.3)

# Спектр после дифференцирования
spectrum_diff_result = out_wave_diff.make_spectrum()
spectrum_diff_result.plot(high=1000, ax=axes[1])
axes[1].set_title('Спектр после дифференцирования')
axes[1].set_xlabel('Частота (Гц)')
axes[1].set_ylabel('Амплитуда')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

**Комментарий:** 
- Метод `diff` работает во временной области и дает четкий прямоугольный сигнал
- Метод `differentiate` работает в частотной области и создает артефакты "звона" из-за разрывов
- Для гладких функций оба метода дают схожие результаты, но для функций с разрывами `diff` предпочтительнее
